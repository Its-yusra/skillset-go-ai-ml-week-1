Work through mean, median, variance, standard deviation, covariance, and correlation using small

**Mean**

```python
data = [10, 20, 30, 40, 50]

mean = sum(data) / len(data)

print("Mean:", mean)
```

Mean is the average value of the dataset.

**Median**

```python
import numpy as np

data = [10, 20, 30, 40, 50]

print("Median:", np.median(data))
```

Variance

```python
print("Variance:", np.var(data))
```

**Standard Deviation**

```python
print("Standard Deviation:", np.std(data))
```

**Covariance**

```python
x = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 6, 8, 10])

print(np.cov(x, y))
```

**Correlation**

```python
print(np.corrcoef(x, y))
```

Correlation shows how two variables move together. It does not by itself prove causation.

**Probability**

```python
total = 10
favorable = 4

probability = favorable / total

print("Probability:", probability)
```

**Conditional Probability**

```python
P_A = 0.5
P_B_given_A = 0.8

P_A_and_B = P_A * P_B_given_A

print("P(A and B):", P_A_and_B)
```

**Vector**

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print("Vector A:", a)
print("Vector B:", b)
```

**Dot Product**

```python
dot_product = np.dot(a, b)

print("Dot Product:", dot_product)
```

A dot product produces a weighted sum and is commonly used in linear models and neural-network layers.

**Matrix**

```python
A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])

print("A:")
print(A)

print("B:")
print(B)

print("A + B:")
print(A + B)

print("A × B:")
print(np.dot(A, B))
```

```python
print("Shape of A:", A.shape)
print("Shape of B:", B.shape)
```

**Derivative**

```python
def f(x):
    return x**2
```

For f(x) = x², the derivative describes how quickly the function changes with respect to x.

**Gradient Descent**

Gradient descent is an optimization method.

Parameters produce a loss.

The gradient tells us the direction of change.

The parameters are then updated to reduce the loss.
